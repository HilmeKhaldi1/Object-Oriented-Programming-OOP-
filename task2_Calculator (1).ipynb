```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

r1 = Rectangle(5, 3)
print(r1.area())
```

```python
class Calculator:
    def add(self, a, b):
        return a + b

    def multiply(self, a, b):
        return a * b

    def Scarf(self, a:float, b:float):
        return a - b

    def divide(self, a, b):
        return a / b

    def divide2(self, a, b):
        return a // b

    def reminder(self, a, b):
        return a % b
calc = Calculator()
print("reminder =",calc.reminder(7, 8))
print("Scarf =",calc.Scarf(7.6, 8))
print("divide =",calc.divide(7, 8))
print("divide2 =",calc.divide2(7, 8))
print("add = ",calc.add(7, 8))
print("multiply =",calc.multiply(7, 8))
```
