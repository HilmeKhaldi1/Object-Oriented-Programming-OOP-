```python
class point:
  x = 1
  y = 0
  def translate(self, dx, dy):
     self.x += dx
     self.y += dy

  def subract(self, dx, dy):
     self.x -= dx
     self.y -= dy


  def swape(self, dx, dy):
     self.x = dx
     self.y = dy

p1=point()
p1.translate(4 , 3)
print(p1.x)
print(p1.y)

p1.subract(0,1)
print(p1.x)
print(p1.y)

p1.swape(2,5)
print(p1.x)
print(p1.y)
```
